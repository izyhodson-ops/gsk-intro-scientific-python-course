```python
Assign a few variables of different types (str, int, float, list, dict) and use type() to confirm each
Pick a string and try at least three different string methods (e.g. upper, split, join)
Pick a list and try append, slicing, and negative indexing
Build a dictionary with at least two entries and retrieve a value with both [] and .get()
```

```python
my_str = "Hello, Python!"
my_int = 42
my_float = 3.14
my_list = [1, 2, 3, "four", 5.0]
my_dict = {"name": "Isabella", "role": "programmer"}

print(type(my_str))    # <class 'str'>
print(type(my_int))    # <class 'int'>
print(type(my_float))  # <class 'float'>
print(type(my_list))   # <class 'list'>
print(type(my_dict))   # <class 'dict'>

print(my_str.upper())             # HELLO, PYTHON!
print(my_str.split(", "))         # ['Hello', 'Python!']
print("-".join(["a", "b", "c"]))  # a-b-c
print(my_str.replace("Hello", "Hi"))  # Hi, Python!

my_list.append("six")
print(my_list)          # [1, 2, 3, 'four', 5.0, 'six']
print(my_list[1:4])     # [2, 3, 'four']  (slice)
print(my_list[-1])      # 'six'           (negative indexing)
print(my_list[-3:])     # ['four', 5.0, 'six']       (negative indexing)

my_dict["language"] = "Python"
print(my_dict)
print(my_dict["name"])                              # Isabella
print(my_dict.get("role"))                           # programmer
print(my_dict.get("missing_key", "default value"))   # default value
```

```python
3. Control Flow

Write an if/elif/else statement that checks something about one of the variables you created above
Write a for loop that builds a new list from an existing one (e.g. squaring each number, or uppercasing each string)



my_int = 100
my_list = [1, 2, 3, "four", 5.0, "six"]

if my_int < 0:
    print("negative")
elif my_int == 0:
    print("zero")
else:
    print("positive")

squared = []
for item in my_list:
    if isinstance(item, (int, float)):
        squared.append(item ** 2)
    else:
        squared.append(item)

print(squared)
# -> [1, 4, 9, 'four', 25.0, 'six']
```

```python
4. Use at least three different built-in functions (e.g. len(), max(), min(), any(), all()) on your own data
In a Markdown cell, answer: how is all() different from any()?

my_list = [1, 2, 3, "four", 5.0, "six"]
nums = [1, 2, 3, 5.0]

print(len(my_list))    # 6    --> number of items in my_list
print(max(nums))       # 5.0  --> largest value in nums
print(min(nums))       # 1    --> smallest value in nums
print(sum(nums))       # 11.0 --> total of nums
print(sorted(nums))    # [1, 2, 3, 5.0] -> sorted list of "nums"

checks = [n > 0 for n in nums]
print(any(checks))     # True -> at least one value is > 0
print(all(checks))     # True -> every value is > 0
```

```python
"all()" differs from "any()" by every requirement within a set/list must be try to trigger  "all()", 
whereas "any()" is more promiscuous and can be triggered by at least 1 item in the set/list being true.
```

```python
def classify_and_square(numbers):
    result = []
    for n in numbers:
        if n % 2 == 0:
            result.append(n ** 2)
        else:
            result.append(n - 1)
    return result
print(result)  # [0, 4, 2, 16, 4, 36]
```
