## Control Flow

### Truthiness

Every object in Python has a boolean value, even if it's not a `bool`. Falsy by default: `None`, `False`, zero of any numeric type, and empty sequences/collections (`''`, `()`, `[]`, `{}`, `set()`, `range(0)`). Everything else is truthy.

### `if` / `elif` / `else`

Python uses indentation instead of parentheses/braces to mark scope. `elif` only runs if the `if` above it failed; `else` runs if nothing matched.

```python
if 5 > 6:
    print("the universe is broken")
elif len("cat") > len("dog"):
    print("uh oh, something is wrong in the world")
else:
    print("math is still mathing")
```

```python
x = 7
if x % 2 == 0:
    print("even")
elif x < 0:
    print("negative")
else:
    print("odd")
```

### `for` Loops

Iterables are objects you can iterate over — lists, dictionaries, and more.

```python
numbers = [1, 2, 3, 4, 5]

for number in numbers:
    print(number)
```

```python
numbers_squared = []
for number in numbers:
    numbers_squared.append(number ** 2)

numbers_squared
```

```python
total = 0
for n in [1, 2, 3, 4, 5]:
    total = total + n

print(total)
```

```python
len([1, 2, 3])
```

```python
max(4, 7, 2)
```

```python
min(4, 7, 2)
```

```python
any([False, False, True])
```

```python
all([True, True, False])
```

## Functions

Functions let you reuse a block of code. Defined with `def`, named in `snake_case`.

```python
def print_hello_world():
    print("Hello world!")

print_hello_world()
```

`f"..."` is an f-string — it lets you embed a variable directly inside a string. `print()` shows something on screen; `return` lets you store the result for later use.

```python
def greet(person):
    return f"Hello {person}!"

message = greet("Teon")
message
```

## Bringing It All Together

Combine control flow, built-ins, and a function.

```python
def classify_numbers(numbers):
    result = []
    for n in numbers:
        if n % 2 == 0:
            result.append("even")
        else:
            result.append("odd")
    return result

classify_numbers([4, 7, 2, 9, 1, 6])
```
